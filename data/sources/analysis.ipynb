```python
import pandas as pd
```

```python
df = pd.read_csv("data/offshore_wind_cable_market.csv")
df
```

```python
df["Calculated_YoY_%"] = (
    df["Market_Size_USD_Billion"]
    .pct_change()
    * 100
)

df["Calculated_YoY_%"] = df["Calculated_YoY_%"].round(2)

df
```

```python
start_value = df.loc[df["Year"] == 2023, "Market_Size_USD_Billion"].iloc[0]
end_value = df.loc[df["Year"] == 2027, "Market_Size_USD_Billion"].iloc[0]

years = 4

cagr_2023_2027 = ((end_value / start_value) ** (1 / years) - 1) * 100

print(f"2023–2027 CAGR: {cagr_2023_2027:.2f}%")
```

```python
summary = df[
    [
        "Year",
        "Market_Size_USD_Billion",
        "Classification",
        "Calculated_YoY_%"
    ]
].copy()

summary
```

## Key Findings

- Global offshore wind cable market size increases from USD 4.049B in 2023 to USD 7.046B in 2027.
- 2025 is treated as the Base Year.
- 2026 and 2027 are forecast values from the same source.
- The strongest annual growth in the selected period occurs between 2025 and 2026.
- The calculated 2023–2027 CAGR is approximately 14.86%.
- The source-reported long-term CAGR for 2025–2032 is 15.96%.
- This dataset is used as a proxy for subsea power cable demand growth, not as a complete measure of the entire subsea power cable market.
